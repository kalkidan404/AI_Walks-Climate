```python
import numpy as np

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("../data/raw_data/tanzania (1).csv")
```

```python
df["country"]="Tanzania"
df["date"]=pd.to_datetime(df["YEAR"]*1000+df["DOY"], format="%Y%j")
df["Month"]=df["date"].dt.month
```

```python
df.replace(-999, np.nan, inplace=True)
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
missing_percentage=df.isna().mean()*100
```

```python
missing_percentage[missing_percentage>5]
```

```python
cols = [
    "T2M",
    "T2M_MAX",
    "T2M_MIN",
    "PRECTOTCORR",
    "RH2M",
    "WS2M",
    "WS2M_MAX"
]
```

```python
Z_score=(df[cols]-df[cols].mean())/df[cols].std()
```

```python
outlier_rows=(Z_score.abs()>3).any(axis=1)
```

```python
outlier_rows.sum()
```

```python
row_missing_percentage = df.isna().mean(axis=1) * 100
```

```python
df=df[row_missing_percentage<=30]
```

```python
df.to_csv("../data/processed_data/Tanzania_clean.csv", index=False)
```
